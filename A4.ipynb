```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load data
df = pd.read_csv("Heart.csv")

# Shape
print("Shape:", df.shape)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Data types
print("\nData Types:")
print(df.dtypes)

# Zero values
print("\nZero Values:")
print((df == 0).sum())

# Mean age
print("\nMean Age:", df["Age"].mean())

# Select columns
data = df[["Age", "Sex", "ChestPain", "RestBP", "Chol"]]

print("\nSelected Data:")
print(data.head())

# Split data
train, test = train_test_split(
    data,
    test_size=0.25,
    random_state=42
)

print("\nTraining Data Shape:", train.shape)
print("Testing Data Shape:", test.shape)

print("\nTraining Data:")
print(train)

print("\nTesting Data:")
print(test)
```
