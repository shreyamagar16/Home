```python
print("===================================")
print(" Welcome to ABC Hospital Support ")
print(" Type 'exit' to end the chat.")
print("===================================")

while True:
    user = input("You: ").lower()

    if user == "exit" or user == "bye":
        print("Bot: Thank you for contacting ABC Hospital. Take care!")
        break

    elif "hello" in user or "hi" in user:
        print("Bot: Hello! Welcome to ABC Hospital. How may I help you?")

    elif "appointment" in user:
        print("Bot: You can book an appointment by calling our reception or visiting the hospital.")

    elif "doctor" in user:
        print("Bot: We have specialists in Cardiology, Orthopedics, Pediatrics, Neurology, and General Medicine.")

    elif "timing" in user or "hours" in user:
        print("Bot: Our hospital is open 24/7. Outpatient (OPD) timings are 9:00 AM to 5:00 PM.")

    elif "emergency" in user:
        print("Bot: Our Emergency Department is available 24 hours a day.")

    elif "fees" in user or "charges" in user:
        print("Bot: Consultation fees vary by doctor. Please contact the reception for details.")

    elif "contact" in user:
        print("Bot: You can contact us at 020-12345678.")

    elif "address" in user or "location" in user:
        print("Bot: ABC Hospital, Main Road, Pune, Maharashtra.")

    elif "thankyou" in user:
        print("Bot: You're welcome! Get well soon.")

    else:
        print("Bot: Sorry, I didn't understand your question. Please ask about appointments, doctors, timings, emergency, fees, contact, or address.")
```
